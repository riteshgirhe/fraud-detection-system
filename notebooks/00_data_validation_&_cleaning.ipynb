```python
df=pd.read_csv("../data/raw/raw_data.csv")
```

```python
print("Dataset shape:",df.shape)
print("*"*100)
print("Column names:",df.columns.tolist())
print("*"*100)
print("Data types:",df.dtypes)
print("*"*100)
print("Missing values:",df.isnull().sum())
print("*"*100)
print("Duplicate rows:",df.duplicated().sum())
print("*"*100)
```
