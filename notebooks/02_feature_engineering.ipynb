```python
import pandas as pd
```

```python
df=pd.read_csv("../data/processed/fraud_cleaned.csv")
```

```python
# 7.1 Balance Difference Features
df["origin_balance_change"] = (
    df["oldbalanceOrg"] - df["newbalanceOrig"]
)

df["destination_balance_change"] = (
    df["newbalanceDest"] - df["oldbalanceDest"]
)
```

```python
# 7.2 Balance Error Features
df["origin_balance_error"] = (
    df["oldbalanceOrg"]
    - df["amount"]
    - df["newbalanceOrig"]
)

df["destination_balance_error"] = (
    df["oldbalanceDest"]
    + df["amount"]
    - df["newbalanceDest"]
)
```

```python
# 7.3 Zero-Balance Indicators
df["origin_old_balance_zero"] = (
    df["oldbalanceOrg"] == 0
).astype(int)

df["origin_new_balance_zero"] = (
    df["newbalanceOrig"] == 0
).astype(int)

df["destination_old_balance_zero"] = (
    df["oldbalanceDest"] == 0
).astype(int)

df["destination_new_balance_zero"] = (
    df["newbalanceDest"] == 0
).astype(int)
```

```python
# 7.4 Amount-to-Balance Ratio
df["amount_to_origin_balance"] = (
    df["amount"] /
    (df["oldbalanceOrg"] + 1)
)
```

```python
# 7.5 Time Features
df["hour"] = df["step"] % 24
```

```python
df["day"] = df["step"] // 24
```

```python
df["is_night"] = (
    (df["hour"] < 6) |
    (df["hour"] >= 22)
).astype(int)
```
